## 1. Descripción del problema

Para el problema de clasificación seleccionamos un dataset que contiene las características fisico-químicas de muestras de vinos, en sus variantes tinto y blanco, de la marca portuguesa de vinos "Vinho Verde". El mismo fue creado con el fin de construir modelos de clasificación multivariable para predicción de calidad de vino. El objetivo, mediante la construcción de un modelo de clasificación, es obtener a partir de las características fisico-químicas (entre las que se encuentra la acidez, concentración de ácido cítrico, pH, etc.) a qué variante de vino se trata, blanco o tinto. Es por ésto que a través de un Análisis Exploratorio y posterior Transformación de datos, se buscará adaptar el dataset para tal fin.

El dataset contiene los siguientes datos:

- fixed_acidity. Acidez total: es la cantidad de acidez, producto de la presencia de diferentes ácidos en el vino (tipo: flotante).

- volatile_acidit. Acidez volátil. Cantidad de ácidos volátiles en el vino (tipo: flotante).

- citric_acid. Ácido cítrico. Cantidad de ácido cítrico presente, que contribuye a la acidez total (tipo: flotante).

- residual_sugar. Azúcar residual: cantidad de azúcar luego de la fermentación (tipo: flotante).

- chlorides. Cloruros: Cantidad de iones cloruro, que indica la presencia de sales en el vino (tipo: flotante).

- free_sulfur_dioxide. Dióxido de azufre libre: cantidad de dióxido de azufre libre en el vino, usado como conservante (tipo: flotante).

- total_sulfur_dioxide. Dioxido de azufre total: cantidad total de dióxido de azufre, incluído libre y en compuesto (tipo: flotante).

- density. Densidad: densidad del vino, relacionada al alcohol y el contenido de azúcar (tipo: flotante).

- pH. Nivel de pH en el vino, indica su acidez (tipo: flotante).

- sulphates. Sulfatos: cantidad de iones sulfato presentes en el vino. Contribuye a su sabor y conservación (tipo: flotante).

- alcohol. Alcohol: porcentaje de contenido de alcohol en el vino (tipo: flotante).

Dicho dataset se encuentra alojado en el sitio Kaggle y se compone de dos archivos, cada uno de los cuales aloja muestras de datos de cada variante.

Link al dataset: https://www.kaggle.com/datasets/joebeachcapital/wine-quality/data

Éste dataset fue seleccionado por el hecho que presenta casi en su totalidad, variables numéricas, que son con las que mejor trabajan los modelos de Machine Learning. Además, se encuentra listo (en lo que respecta a limpieza y estructura de datos) para trabajar en la construcción de modelos. ### 2.2 Limpieza y consolidación de datos

Nuestro dataset de clasificación se halla separado en dos tablas: una contiene las muestras que pertenecen a la clase *blanco*, mientras que la otra contiene las que pertenecen a la muestra *tinto*. Con la utilización de la función **.concat()** de Pandas, uniremos ambos datasets.

Dichas tablas no poseen de forma explicita a qué clase pertenecen los atributos, por lo que deberemos agregarlo de forma manual. Definiremos a las muestras de vino blanco como *class = $0$*, mientras que las muestras de vino tinto las definiremos como *class = $1$*.

Finalmente, prescindiremos de la columna *quality* (que expresa el puntaje o calidad de la muestra de vino) ya que es la columna que contiene la clase objetivo en caso que el dataset sea untilizado para un modelo de clasificación multivariable (nuestro objetivo es construir un modelo de Regresión Logística para clasificación binaria).

```python
wine_quality_white['class'] = 0
wine_quality_red['class'] = 1

wine_quality = pd.concat([wine_quality_white, wine_quality_red], ignore_index=True)
wine_quality = wine_quality.drop(['quality'], axis=1)
print("Dataset 'winequality'")
wine_quality
```

Además, puede provocar que el modelo utilice información sobre la distribución de los datos de prueba durante el entrenamiento, lo que genera métricas de éxito engañosas.

Realizaremos una división 70-30 (70% entrenamiento, 30% prueba) utilizando la función **train_test_split()** de la librería *scikit-learn*:

```python
# Importamos las funciones necesarias de scikit-learn:
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Creamos la particiones:
X_clas_train, X_clas_test, y_clas_train, y_clas_test = train_test_split(X_clas, y_clas, test_size=0.30, random_state=15)

# Entrenamos el modelo:
modelo_clas = LogisticRegression(max_iter=1000)
modelo_clas.fit(X_clas_train, y_clas_train)

# Mostramos por pantalla el rendimiento:
resultado_accuracy = modelo_clas.score(X_clas_test, y_clas_test)
print(f"Promedio o Accuracy: {resultado_accuracy*100.0:.2f}%")
```

La métrica *'accuracy'* para evaluar un modelo no es la ideal cuando las clases están desbalanceadas. Por otro lado, un modelo debe mantener su precisión en un rango de $90$ a $95$ para lograr un equilibrio óptimo entre utilidad práctica, confianza estadística y prevención de errores (como el *sobreajuste* ). Por ello, evaluaremos nuevamente el modelo, pero con el score *Kappa*:

```python
#Importamos la clase que contiene el score
from sklearn.metrics import cohen_kappa_score

# Realizamocs la evaluación:
predicciones = modelo_clas.predict(X_clas_test)
resultado_cohen = cohen_kappa_score(y_clas_test, predicciones)

print(f"Cohen Kappa Score: {resultado_cohen*100.0:.2f}%")
print(f"Desviación Estándar {resultado_cohen.std()*100.0:.2f}%")
```

A pesar de obtener un porcentaje de evaluación bastante confiable, es un modelo creado con datos sin procesar, lo que puede derivar en modelos de bajo rendimiento, predicciones erróneas y sesgos por la presencia de ruidos y valores atípicos. Para evitar éstos inconvenientes, procederemos al procesamiento de los datos.

### 3.2 Métodos de preprocesamiento

Para transformar nuestros datos, vamos a utiliza dos métodos.

#### Transformación *Box-Cox*

En el apartado de Análisis Exploratorio de Datos, observamos que *residual sugar* posee sesgo positivo y *alcohol* posee una distribución aplanada, por lo que utilizaremos ésta transformación para convertir los datos *no* normales en una distribución similar a una *gaussiana*.

La transformación de *Box-Cox* es una familia de técnicas estadísticas que convierte datos no normales en una distribución más cercana a la normal. Mediante un parámetro de potencia ($λ$), estabiliza la varianza y corrige la asimetría, permitiendo que los datos cumplan los requisitos de modelos predictivos.

Seleccionaremos las ubicaciones en la partición de entrenamiento para aislar los valores correspondientes a dicho atributo y realizar la transformación. La columna *residual sugar* ocupa el 4to lugar (índice = 3) en el dataset original, mientras que *alcohol* ocupa el 11mo lugar (índice = 10).

```python
wine_quality_final[['residual sugar', 'alcohol']].head(10)
```

```python
# Importamos función de scikit-learn
from sklearn.preprocessing import PowerTransformer

# Tomamos los datos que corresponden a los atributos a transformar
# de la partición de entranamiento
atributos_clas_train = X_clas_train[:,[3,10]]

# Aplicamos Transformación de potencia 'Box-Cox':
transformer = PowerTransformer(method='box-cox')
atributos_clas_train_tf = transformer.fit_transform(atributos_clas_train)

# De array de NumPy a DataFrame
atributos_clas_train_tf_df = pd.DataFrame(atributos_clas_train_tf,
                                       columns=['residual sugar', 'alcohol'])
atributos_clas_train_tf_df.head()
```

Ya transformados los valores del set de entrenamiento, haremos lo mismo en el set de prueba, utilizando directamente la función **.transform()**.

```python
# Tomamos los datos que corresponden en la partición de prueba
atributos_clas_test = X_clas_test[:,[3,10]]

# Transformamos la partición de prueba, pero solamente mediante la
# función transform()
atributos_clas_test_tf = transformer.transform(atributos_clas_test)
# De array de NumPy a DataFrame
atributos_clas_test_tf_df = pd.DataFrame(atributos_clas_test_tf,
                                      columns=['residual sugar', 'alcohol'])
atributos_clas_test_tf_df.head()
```

Con los datos transformados, crearemos un DataFrame para visualizar los datos y comparar:

### 3.3 Remuestreo de los datos

El remuestreo (o resampling) es un paso importante en la construcción de un modelo de Machine Learning, ya que su fin es estimar la precisión de los modelos, calcular errores estándar e intervalos de confianza sin asumir distribuciones teóricas complejas. Consiste en extraer repetidamente submuestras de un conjunto de datos original para obtener información más robusta.

El K-fold Cross Validation (Validación Cruzada de $K$ iteraciones) es una técnica de Machine Learning utilizada para evaluar el rendimiento de un modelo y evitar el sobreajuste. Consiste en dividir el conjunto de datos original en \($K$) subconjuntos iguales (llamados $folds$). El modelo se entrena con $K-1$ folds y se prueba con el fold restante. Este proceso se repite $K$ veces, de modo que cada fold actúa exactamente una vez como conjunto de prueba. Finalmente, se promedian los resultados (como la precisión o el error) para obtener una estimación sólida del rendimiento general del modelo.

Ya que el dataset 'winequality' posee una gran cantidad de muestras, usaremos el método *K-Fold Cross Validation* sin repetir. El mismo posee un desbalance de clases, por lo que utilizaremos *StratifiedKFold*, una variante del algoritmo que divide el conjunto de datos en *$k$* particiones garantizando que cada partición mantenga la misma proporción de ejemplos por clase que nuestro conjunto original, además como método de scoring *Cohen's Kappa*.

Comezaremos las pruebas con un tamaño de partición $k = 5$:

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import make_scorer

# Definimos la cantidad de particiones a generar
nro_k = 5
# Seed o semilla para asegurar reproducibilidad
semilla = 42
# Generación de particiones
stkfold = StratifiedKFold(n_splits=nro_k, random_state=semilla, shuffle=True)

# Función para utilizar scoring no nativo
kappa_scorer = make_scorer(cohen_kappa_score)

# Definimos el número de iteraciones máximas hasta encontrar
# el mejor rendimiento de nuestro modelo
modelo_clas_uno = LogisticRegression(max_iter=1000)
resultado_kfold_5 = cross_val_score(modelo_clas_uno, X_clas_train_estandarizado,
                                    y_clas_train, cv=stkfold, scoring=kappa_scorer)
```

```python
print(f"Promedio o Accuracy: {resultado_kfold_5.mean()*100.0:.2f}%")
print(f"Desviación Estándar {resultado_kfold_5.std()*100.0:.2f}%")
```

Y ahora lo haremos con un tamaño $k = 10$:

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

# Definimos la cantidad de particiones a generar
nro_k = 10
# Seed o semilla para asegurar reproducibilidad
semilla = 42
# Generación de particiones
stkfold = StratifiedKFold(n_splits=nro_k, random_state=semilla, shuffle=True)

# Definimos el número de iteraciones máximas hasta encontrar
# el mejor rendimiento de nuestro modelo
resultado_kfold_10 = cross_val_score(modelo_clas_uno, X_clas_train_estandarizado,
                                     y_clas_train, cv=stkfold, scoring=kappa_scorer)
```

```python
print(f"Promedio o Accuracy: {resultado_kfold_10.mean()*100.0:.2f}%")
print(f"Desviación Estándar {resultado_kfold_10.std()*100.0:.2f}%")
```

La precisión se mantuvo similar en ambos casos y en los límites ideales. Por otro lado, hubo un leve aumento del desvío estandar, superior al 1% con *$k$ = 10*.

### 3.4 Evaluación de Algoritmos

#### A. Taxonomía Lineal: Regresión Logística

El rendimiento de un modelo de machine learning indica su eficacia para predecir o tomar decisiones con datos nuevos. Se mide mediante métricas cuantitativas — tales como exactitud, precisión o error cuadrático medio — y es fundamental para garantizar que el algoritmo funcione correctamente en entornos reales. Para los casos de modelos de clasificación, y en especial Regresiones Logísticas, se utilizan técnicas y métricas basados en la Matriz de Confusión, como la Exactitud (Accuracy) y la Curva ROC-AUC.

- Área bajo la curva ROC

```python
# Generamos las particiones para la validación cruzada
# y puntuar el modelo
stkfold = StratifiedKFold(n_splits=nro_k, random_state=semilla, shuffle=True)

array_result = cross_val_score(modelo_clas_uno, X_clas_train_estandarizado, y_clas_train, cv=stkfold, scoring='roc_auc')
print(f"Área Bajo la Curva: {array_result.mean()*100.0:.2f}%")
print(f"Desviación Estándar {array_result.std()*100.0:.2f}%")
```

En nuestro primer modelo el valor de Área ROC es igual a $0.99$, lo cual significa que dicho modelo tiene un 99% de probabilidades de clasificar de manera correcta una muestra como vino blanco (que en nuestro caso es la clase positiva o $0$). El inconveniente es que éste método de puntuación no funciona correctamente con muestras con clases desbalanceadas, por lo que para éstos casos es más adecuado utilizar **Reportes de Clasificación**.

```python
from sklearn.metrics import classification_report

# Creamos el modelo de Regresión Logística
modelo_clas_uno = LogisticRegression(max_iter=1000)

# Entrenamiento del modelo
modelo_clas_uno.fit(X_clas_train_estandarizado, y_clas_train)
predicciones_modelo_clas_uno = modelo_clas_uno.predict(X_clas_test_estandarizado)
reporte_clasificacion_clas_uno = classification_report(y_clas_test, predicciones_modelo_clas_uno)

print("        --- Reporte de clasificación Árbol de Decisión ---\n")
print(reporte_clasificacion_clas_uno)
```

Con un *accuracy* del 99%, el resultado es óptimo. Sin embargo, los reportes de clasificación deben complementarse con validación cruzada para verificar si estamos en presencia de un sobreajuste del modelo. Para éste modelo obtuvimos un rendimiento del 96%, lo que muestra un buen desempeño con éstos datos.

#### B. Taxonomía No Lineal: CaRT

El algoritmo de Taxnomía No Lineal seleccionado fue el Árbol de Decisión. Los Árboles de Decisión (o **CaRT**, siglas en inglés de Árboles de Regresión y Clasificación ) son modelos de aprendizaje supervisado no paramétrico. Crean un modelo que predice el valor o la clase de una variable objetivo mediante el aprendizaje de reglas de decisión simples inferidas a partir de las características de los datos. Su estructura comienza con un nodo raíz (la característica más importante) y se ramifica hacia abajo haciendo preguntas condicionales. Allí radica su facilidad tanto para visualizarlos y como para entenderlos. Poseen la ventaja de requerir poco procesamiento de los datos, ya que no se basan en distancias geométricas ni utilizan gradientes, pero son propensos a caer en el sobreajuste.

Crearemos y evaluaremos un Árbol de Decisión para luego realizar el remuestreo correspondiente para evaluar su rendimiento:

```python
# Importamos el Árbol de Decisión
from sklearn.tree import DecisionTreeClassifier

# Creamos el Árbol
modelo_dt_uno = DecisionTreeClassifier(random_state=42)
```

```python
resultado_kfold_5_dt_est = cross_val_score(modelo_dt_uno, X_clas_train_estandarizado,
                                           y_clas_train, cv=stkfold, scoring = kappa_scorer)
print(f"Promedio o Accuracy: {resultado_kfold_5_dt_est.mean()*100.0:.2f}%")
print(f"Desviación Estándar {resultado_kfold_5_dt_est.std()*100.0:.2f}%")
```

Ambos modelos fueron entrenados con el mismo set de datos, el cual fue preprocesado para poder entrenar el algoritmo de Regresión Logística. Eso marca una ventaja para el Árbol de Decisión, ya que no es necesario dicho paso.

Como parámetro de scoring se utilizó el *Cohen's Kappa* al igual que la Regresión Logística debido al desbalance de clases, ofreciendo una evaluación más precisa del modelo.

En el modelo de Regresión Logística habiamos obtenido un rendimiento promedio del 96.99%, con una desviación estándar 0.83%. Los valores obtenidos con el Árbol de Decisión son similares: el desvío estandar es mayor en éste último, pero manteniéndose en bajo, lo que señala buena estabalidad. Es por ello que deducimos un desempeño óptimo en ambos modelos.

Generamos un reporte de clasificación para el Árbol de Decisión y lo comparamos con el reporte generado para el modelo de Regresión:

```python
# Entrenamos el Árbol de Decisión
modelo_dt_uno.fit(X_clas_train_estandarizado, y_clas_train)

# Realizamos las predicciones
predicciones_modelo_dt_uno = modelo_dt_uno.predict(X_clas_test_estandarizado)

# Generamos Reporte de Clasificación
reporte_clasificacion_dt = classification_report(y_clas_test, predicciones_modelo_dt_uno)

print("        --- Reporte de clasificación Árbol de Decisión ---\n")
print(reporte_clasificacion_dt)
print("\n      --- Reporte de clasificación Regresión Logística ---\n")
print(reporte_clasificacion_clas_uno)
```

## 4. Feature Selection/Feature Importance

La selección de características (Feature Selection) es el proceso aplicar distintas técnicas para elegir las variables más útiles y relevantes del conjunto de datos para entrenar un modelo. Su objetivo principal es mejorar la precisión del modelo, reducir el tiempo de entrenamiento y evitar el sobreajuste (overfitting) al eliminar datos irrelevantes.

Despues de aplicar distintos métodos de preprocesamiento de datos y evaluación de algoritmos, aplicaremos el Feature Selection para disminuir lo que a priori significaria un problema de overtfitting de nuestro modelo.

### 4.1 Características altamente correlacionadas

A la hora de construir modelos de Machine Learning, la presencia de atributos altamente correlacionados (multicolinealidad) reduce la precisión y la estabilidad, ya que se dificulta aislar el impacto individual de cada variable en la predicción final.

Mientras que los atributos de alta correlación con la clase objetivo pueden ayudar a la construcción de un buen modelo, la correlación entre atributos puede introducir ruido y redundancia, lo cual lo empeora. Para evitar ésto, eliminaremos las características altamente correlacionadas para mejorar la estabilidad y la interpretabilidad del modelo.

## 5. Análisis de Componentes Principales(PCA)

### 5.1 Aplicación del método

El PCA (Análisis de Componentes Principales) es un algoritmo de aprendizaje automático no supervisado utilizado para reducir la dimensionalidad de sets de datos. Permite comprimir la información eliminando variables redundantes en un número menor de variables nuevas llamadas *componentes principales*, lo que simplifica los cálculos, reduce el ruido y el sobreajuste y facilita la visualización gráfica de datos complejos. Estas nuevas variables son combinaciones lineales de las originales y logran conservar la mayor cantidad de información posible (varianza).

Importamos la función desde la librería de *scikit-learn* y en el parámetro *n_components* dejaremos que el modelo determine la cantidad mínima de parámetros que deberá retener el 90% de la varianza ($0.90$), aplicando éste método sobre nuestros datos estandarizados:

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=0.90)

fit_pca = pca.fit(X_clas_train_estandarizado)
```

El algoritmo calcula que el mínimo de características son 7. Detallamos el porcentaje de varianza para cada PC (ya que la PCA la maximiza):

```python
fit_pca.explained_variance_ratio_
```

Las siguientes filas contienen los vectores propios que definen cómo se combinan linealmente las variables originales para formar los nuevos componentes principales.

```python
fit_pca.components_
```

Aplicamos **.fit()** y **.transform()** para PCA en el set de entrenamiento:

```python
X_clas_train_pca = fit_pca.transform(X_clas_train_estandarizado)

df_clas_train_pca = pd.DataFrame(data=X_clas_train_pca, columns=['PC1', 'PC2', 'PC3', 'PC4', 'PC5', 'PC6', 'PC7'])
df_clas_train_pca.head(5)
```

Y realizamos lo mismo con el set de prueba:

```python
X_clas_test_pca = fit_pca.transform(X_clas_test_estandarizado)
X_clas_test_pca
```

Vemos que con ésta técnica, la precisión del modelo aumento y mejoró bastante. Podriamos plantear la diferencia en el número de características ideales para la predicción de la clase como una variable que influye en el desempeño del modelo.

### 5.2 Gráfico de codo

### 5.3 ¿Qué se busca con *Feature Importance*?

Con Análisis de Componentes Principales (PCA), el objetivo es evaluar la importancia de las características para identificar qué variables originales tienen mayor influencia en los nuevos componentes principales. Se basa en dos métricas clave:

- Los pesos o coeficientes que relacionan cada variable original con los Componentes Principales: a mayor magnitud absoluta, indica que la variable original aporta más información a un componente en específico.

- El aporte a la varianza desde los componentes principales, los cuales acumulan la mayor cantidad de información (por ejemplo, el 95% o 99% de la varianza total). Las variables originales que más pesan en estos primeros componentes son consideradas las más "importantes".

Debemos remarcar que tanto aquí como en *Feature Selection*, "hemos dejado" al algoritmo en cuestión (Selección Univariable en *Feature Selection* y PCA en *Feature Importance*) que calcule la cantidad de atributos "ideales" para la construcción de nuestro modelo.

Con PCA hemos obtenido un rendimiento del 95% y un desvío estandar cecano al 1,0% en validación cruzada, métricas ideales en comparación con el método anteriormente aplicado.
